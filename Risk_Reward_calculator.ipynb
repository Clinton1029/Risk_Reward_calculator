```python
import ipywidgets as widgets
from IPython.display import display, HTML

# ==========================
# TITLE
# ==========================

display(HTML("""
<h1 style="text-align:center;color:#00C853;">
📈 FOREX RISK CALCULATOR
</h1>

<h4 style="text-align:center;color:gray;">
EUR/USD • GBP/USD
</h4>
"""))

# ==========================
# INPUTS
# ==========================

pair = widgets.Dropdown(
    options=["EUR/USD", "GBP/USD"],
    value="EUR/USD",
    description="Pair:",
    layout=widgets.Layout(width="250px")
)

risk_pips = widgets.FloatText(
    value=3.0,
    description="Risk Pips:",
    layout=widgets.Layout(width="250px")
)

lot_size = widgets.FloatText(
    value=0.04,
    description="Lot Size:",
    layout=widgets.Layout(width="250px")
)

rr = widgets.FloatText(
    value=3.0,
    description="R:R:",
    layout=widgets.Layout(width="250px")
)

calculate_btn = widgets.Button(
    description="CALCULATE",
    button_style="success",
    icon="calculator",
    layout=widgets.Layout(width="180px", height="45px")
)

clear_btn = widgets.Button(
    description="CLEAR",
    button_style="danger",
    icon="trash",
    layout=widgets.Layout(width="180px", height="45px")
)

output = widgets.Output()

# ==========================
# FUNCTIONS
# ==========================

def calculate(b):

    with output:

        output.clear_output()

        try:

            pip_value = lot_size.value * 10

            risk_usd = pip_value * risk_pips.value

            tp_pips = risk_pips.value * rr.value

            profit = tp_pips * pip_value

            display(HTML(f"""

            <div style="
            background:#1e1e1e;
            color:white;
            padding:25px;
            border-radius:15px;
            border:2px solid #00C853;
            width:600px">

            <h2 style="color:#00C853;">
            📊 Trade Summary
            </h2>

            <hr>

            <b>Currency Pair:</b> {pair.value}<br><br>

            <b>Lot Size:</b> {lot_size.value:.2f}<br><br>

            <b>Risk (Pips):</b> {risk_pips.value:.2f}<br><br>

            <b>Risk : Reward:</b> 1 : {rr.value}<br><br>

            <hr>

            <h3 style="color:#FFC107;">
            💰 Risk = ${risk_usd:.2f}
            </h3>

            <h3 style="color:#03A9F4;">
            🎯 Take Profit = {tp_pips:.2f} Pips
            </h3>

            <h3 style="color:#00E676;">
            ✅ Profit = ${profit:.2f}
            </h3>

            </div>

            """))

        except:

            display(HTML("""
            <h3 style='color:red'>
            Please enter valid numbers.
            </h3>
            """))

def clear(b):

    risk_pips.value = 0
    lot_size.value = 0
    rr.value = 0

    with output:
        output.clear_output()

calculate_btn.on_click(calculate)
clear_btn.on_click(clear)

# ==========================
# DISPLAY
# ==========================

display(pair)
display(risk_pips)
display(lot_size)
display(rr)

display(widgets.HBox([calculate_btn, clear_btn]))

display(output)
```
