```python
import ipywidgets as widgets
from IPython.display import display, HTML

display(HTML("<h2 style='color:#58a6ff'>📈 Forex Risk Calculator</h2><p>Automatic updates for EUR/USD & GBP/USD</p>"))

pair=widgets.Dropdown(options=["EUR/USD","GBP/USD"],description="Pair")
risk_pips=widgets.FloatText(value=3.0,description="SL (pips)")
lot_size=widgets.FloatText(value=0.04,description="Lot")
rr=widgets.FloatText(value=3.0,description="R:R")
clear_btn=widgets.Button(description="Reset",button_style="danger")
out=widgets.Output()

def update(change=None):
    with out:
        out.clear_output()
        try:
            pip_value=lot_size.value*10
            risk=risk_pips.value*pip_value
            tp=risk_pips.value*rr.value
            profit=tp*pip_value
            html=f'''
<div style="background:#161b22;color:white;padding:20px;border-radius:12px;width:500px">
<h3 style="color:#58a6ff">Trade Summary</h3>
<table style="width:100%">
<tr><td>Pair</td><td align="right">{pair.value}</td></tr>
<tr><td>Lot Size</td><td align="right">{lot_size.value:.2f}</td></tr>
<tr><td>Risk</td><td align="right">{risk_pips.value:.2f} pips</td></tr>
<tr><td>R:R</td><td align="right">1:{rr.value:.2f}</td></tr>
<tr><td>Pip Value</td><td align="right">${pip_value:.2f}</td></tr>
</table><hr>
<div style="display:flex;gap:10px">
<div style="flex:1;background:#222;padding:10px;border-radius:8px"><b>Risk</b><br><span style="color:#ff5252">${risk:.2f}</span></div>
<div style="flex:1;background:#222;padding:10px;border-radius:8px"><b>TP</b><br><span style="color:#4fc3f7">{tp:.2f} pips</span></div>
<div style="flex:1;background:#222;padding:10px;border-radius:8px"><b>Profit</b><br><span style="color:#00e676">${profit:.2f}</span></div>
</div></div>'''
            display(HTML(html))
        except Exception as e:
            print(e)

for w in [pair,risk_pips,lot_size,rr]:
    w.observe(update,names="value")

def reset(b):
    pair.value="EUR/USD";risk_pips.value=3;lot_size.value=0.04;rr.value=3
clear_btn.on_click(reset)

display(pair,risk_pips,lot_size,rr,clear_btn,out)
update()
```
